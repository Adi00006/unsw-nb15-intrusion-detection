# UNSW-NB15 Real Predictions in Jupyter

Run the saved binary and multiclass models on the real UNSW-NB15 test set and save prediction CSVs.

```python
from pathlib import Path
import os
import sys

ROOT = Path.cwd().resolve()
while not (ROOT / 'src').exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
if not (ROOT / 'src').exists():
    raise FileNotFoundError('Could not find the project root containing src/')
os.chdir(ROOT)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

ROOT, ROOT / 'data', ROOT / 'models'
```

```python
from pathlib import Path
import joblib
import pandas as pd

from src.predict import predict_frame

DATA_DIR = ROOT / 'data'
MODELS_DIR = ROOT / 'models'
OUTPUT_DIR = ROOT / 'predictions'
OUTPUT_DIR.mkdir(exist_ok=True)

TEST_CSV = DATA_DIR / 'UNSW_NB15_testing-set.csv'
BINARY_MODEL = MODELS_DIR / 'binary_real' / 'best_model.joblib'
MULTICLASS_MODEL = MODELS_DIR / 'multiclass_real' / 'best_model.joblib'

print(f'Test CSV: {TEST_CSV}')
print(f'Binary model: {BINARY_MODEL}')
print(f'Multiclass model: {MULTICLASS_MODEL}')
```

```python
test_df = pd.read_csv(TEST_CSV)
binary_artifact = joblib.load(BINARY_MODEL)
binary_out, _ = predict_frame(binary_artifact, test_df)
binary_path = OUTPUT_DIR / 'binary_predictions.csv'
binary_out.to_csv(binary_path, index=False)
print('Binary prediction summary:')
print(binary_out['predicted_class'].value_counts().to_string())
print(f'
Saved binary predictions to {binary_path}')
```

```python
multiclass_artifact = joblib.load(MULTICLASS_MODEL)
multiclass_out, _ = predict_frame(multiclass_artifact, test_df)
multiclass_path = OUTPUT_DIR / 'multiclass_predictions.csv'
multiclass_out.to_csv(multiclass_path, index=False)
print('Multiclass prediction summary:')
print(multiclass_out['predicted_class'].value_counts().to_string())
print(f'
Saved multiclass predictions to {multiclass_path}')
```
